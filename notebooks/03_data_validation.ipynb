```python
total_clean_gross_revenue=cleaned_df["gross_revenue"].sum()
total_clean_return_value=cleaned_df['return_value'].sum()
total_clean_net_revenue=cleaned_df["net_revenue"].sum()

print("Total Cleaned Gross revenue : "+str(total_clean_gross_revenue))
print("Total Cleaned Return value : "+str(total_clean_return_value))
print("Total Cleaned Net Revenue : "+str(total_clean_net_revenue))
```

```python
total_raw_gross_revenue=(raw_df["Quantity"] * raw_df["Price"]).sum()
total_raw_return_value=(raw_df.loc[raw_df["Quantity"] < 0, "Quantity"] * raw_df.loc[raw_df["Quantity"] < 0, "Price"]).abs().sum()
total_raw_net_revenue=total_raw_gross_revenue-total_raw_return_value

print("Total Raw Gross revenue : "+str(total_raw_gross_revenue))
print("Total Raw Return value : "+str(total_raw_return_value))
print("Total Raw Net Revenue : "+str(total_raw_net_revenue))
```
