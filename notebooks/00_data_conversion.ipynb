## Converting source XLSX file in CSV format

```python
import pandas as pd
```

```python
input_file = "../data/raw/online_retail_II.xlsx"
output_file = "../data/processed/online_retail_II_processed.csv"

dfs = pd.read_excel(input_file, sheet_name=None)

df = pd.concat(dfs.values(), ignore_index=True)

df.to_csv(output_file, index=False)
```
